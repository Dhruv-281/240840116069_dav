```python
#Importing all the necessary libraries
import dash
from dash import dcc, html, Input, Output
import plotly.express as px
import pandas as pd
```

```python
#Creating a dataset
Data={
    'Month': ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun'],
    'Sales': [12000, 15000, 18000, 14000, 22000, 25000],
    'Profit': [3000, 4000, 5000, 3500, 6000, 7500]
}
#Converting the dataset into dataframe
df=pd.DataFrame(Data)

#Initializing the Dash application using the 'app' variable
app = dash.Dash(__name__)

# Creating a dashboard layout
app.layout = html.Div([         #app.layout defines the structure of the dashboard and html.Div() merges the components of dashboard into one section
    html.H1("Business Reporting Dashboard"), #It will display the main heading

    html.Label("Select Business Metric:"),
    dcc.Dropdown(      #dcc.Dropdown() creates a dropdown menu.
        id='metric',   #id='metric' uniquely identifies the dropdown
        options=[      #options specifies the available choices: Sales and Profit.
            {'label': 'Sales', 'value': 'Sales'},
            {'label': 'Profit', 'value': 'Profit'}
        ],
        value='Sales', #value='Sales' selects Sales by default.
        clearable=False #clearable=False prevents the selection from being cleared
    ),

    dcc.Graph(id='bar-chart'), #dcc.Graph() displays a chart in the dashboard
    dcc.Graph(id='line-chart') #'id=bar/line-chart identifies the required charts
])

#@app.callback – Connects the dropdown menu with the charts so they can update automatically
@app.callback(
    [Output('bar-chart', 'figure'), #Output('chart-name', 'figure') – Updates the chart with new data
     Output('line-chart', 'figure')],
    Input('metric', 'value')  #Takes the selected value (Sales or Profit) from the dropdown menu.
)
def update_charts(metric):  # Updates charts based on the selected metric
    bar = px.bar(           #Craeting a barchart
        df, x='Month', y=metric,
        title=f'Monthly {metric}'
    )

    line = px.line(         #Creating a line chart
        df, x='Month', y=metric,
        title=f'{metric} Trend',
        markers=True
    )

    return bar, line #Returns both charts

if __name__ == '__main__': #Checks if the file runs directly or not
    app.run(debug=True)    #Starts the dashboard server and enables debugging.
```
