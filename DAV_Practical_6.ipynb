Aim:Create a basic data visualizations

```python
#importing the pandas, numpy and matplotlib library as pd, np and plt respectivelty
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
#Creating a dataset
data = {
    "Month":["Jan","Feb","Mar","Apr","May","June","July","Aug","Sep","Oct","Nov","Dec"],
    "Product":["Laptop","Mobile","Tablet","Laptop","Mobile","Tablet","Laptop","Mobile","Tablet","Laptop","Mobile","Tablet"],
    "Region":["North","South","East","West","North","South","East","West","North","South","East","West"],
    "Sales":[120000,135000,150000,145000,160000,175000,180000,170000,190000,210000,230000,250000],
    "Units_Sold":[12,30,25,14,35,30,18,38,32,21,45,40]
}
#Converting the database into a dataframe
df = pd.DataFrame(data)
#Display the dataframe
df
```

Step-2: Create a linechart

Que: How did the company's sales change from January to December

```python
#Create a linechart using matplotlib library
plt.figure(figsize=(10,5))
#Plotting a graph with month on x-axis and sales on y-axis with marker as "o"
plt.plot(df["Month"],df["Sales"], marker="o")
#Giving the title to the graph
plt.title("Company Sales from January to December")
#Labelling the x-axis as "Month"
plt.xlabel("Month")
#Labelling the y-axis as Sales
plt.ylabel("Sales")
plt.grid(True)
#Displaying the graph
plt.show()
```

Step-3: Create a Barchart

Que:Which product generate how much total sales

```python
#Create a barchart using matplotlib library for the each product's total sales
Product_Sales = df.groupby("Product")["Sales"].sum()#Use the groupby function on Product and Sales
plt.bar(Product_Sales.index,Product_Sales.values)
plt.title("Total Sales by Product")
plt.xlabel("Product") #Labelling the x-axis as "Product"
plt.ylabel("Total Sales") #Labelling the y-axis as "Toatl Sales"
plt.show()
```

Step-4: Create a Histogram

Que:How are the individual sales values distributed

```python
#Create a histogram for sales values distribution using hist()
plt.hist(df["Sales"],bins=10) #bin is used to divide data into several interval
plt.title("Sales Values Distribution")
plt.xlabel("Sales") #Labelling the x-axis as "Sales"
plt.ylabel("Frequency")#Labelling the y-axis as "Frequency"
plt.show()
```

Step-5:Cretae a scatter plot

Que:Is there a realtionship betn the number of units sold and total sales

```python
#Scatter plot for the number of units sold and total sales
plt.scatter(df["Units_Sold"],df["Sales"])
plt.title("Relationship between Units Sold and Sales")
plt.xlabel("Units Sold") #Labelling the x-axis as "Units sold"
plt.ylabel("Sales") ##Labelling the y-axis as "Sales"
plt.show()
```

Step-6:Create a piechart

Que:What percentage of total sales comes from each region

```python
#Create a pie chart on what percentage of total sales comes from each region
Region_Sales = df.groupby("Region")["Sales"].sum() #Use the groupby function on Region and Sales
#autopct is used to get percentage values automatically on each slice of pie
plt.pie(Region_Sales.values,labels=Region_Sales.index,autopct="%1.1f%%")
```
