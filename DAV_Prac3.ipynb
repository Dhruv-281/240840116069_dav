```python
#importing the pandas and the numpy library
import pandas as pd
import numpy as np
#Creating the database
data={
    "Emp_ID":[101, 102, 103, 104, 105, 106, 107, 108],
    "Name":["Rahul","Priya","Amit","Riya","Karan","Neha","Vivek","Pooja"],
    "Department":["IT","HR","IT","Finance","HR","Finance","IT","HR"],
    "City":["Surat","Navsari","Surat","Valsad","Surat","Navsari","Valsad","Surat"],
    "Salary":[45000,38000,52000,48000,40000,51000,60000,42000],
    "Experience":[2,1,5,4,2,6,8,3]
}
#Creating a datafarme of the database
df=pd.DataFrame(data)
```

Step-1

```python
#Displaying the dataframe which we have created
df
```

Step-2

2.1 Display emp from it dept

```python
#Displaying the employees from IT department
df[df["Department"]=="IT"]
```

2.2 Display employees with salary greater than 50000

```python
#Displaying the employees with salary more than 50000
df[df["Salary"]>50000]
```

2.3 Display employees fron Surat

```python
#Displaying the employees belongs Surat
df[df[("City")]=="Surat"]
```

2.4 Display employees from IT department and having salary more than 50000

```python
#Displaying the employees from IT department with salary more than 50000
df[(df["Department"]=="IT")&(df["Salary"]>50000)]# We have used AND operator
```

Step-3 Sort the data

3.1 Sort salaries in ascending order

```python
#Sorting the values using the sort() in ascending order
df.sort_values('Salary', ascending=True)
```

3.2 Sort salaries in descending order

```python
#Displaying salaries in descending using ascending=False
df.sort_values('Salary', ascending=False)
```

3.3 Sort employees by Department and salary

```python
# Sort department in ascending order then set salary fron ach department in descending order
df.sort_values(['Department', 'Salary'], ascending=[True, False])
```

Step-4 Perform aggregation

4.1 Calculate total salary

```python
#Using sum() calculate all employees salary
df["Salary"].sum()
```

4.2 Calculate average salary

```python
#Get the average salary using mean()
df["Salary"].mean()
```

4.3 Find maximum salary

```python
# Get maximum salary using max()
df["Salary"].max()
```

4.4 Find the minimum salary

```python
#Get the minimum salary using min()
df["Salary"].min()
```

4.5Count the employees

```python
#Get the count of total employees using the count() on employees id
df["Emp_ID"].count()
```

Step-5 Perform groupby operations

5.1 Average salary by department

```python
#Group employees by department
#Get average using mean()
df.groupby('Department')['Salary'].mean()
```

5.2 Total salary by department

```python
# Get total salary paid by each department
df.groupby('Department')['Salary'].mean()
```

5.3 Maximum salary by department

```python
# Get maximum salary by deaparyment using max()
df.groupby('Department')['Salary'].max()
```

5.4 Count of employees by department

```python
# Get count of employees in department
df.groupby('Department')['Emp_ID'].count()
```

Average salary by city

```python
# Group by employees by city and use mean()
df.groupby('City')['Salary'].mean()
```

Step-6 Multiple aggregate function

```python
# Apply the multiple statistical functions to the slary column
df.groupby('Department')['Salary'].agg(['mean','sum','max','min','count'])
```

```python
#Convert the dataframe into csv file
df.to_csv("employees_data.csv",index=False)
print("Csv file saved successfully.")
```
