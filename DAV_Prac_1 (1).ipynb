Step-1: Import the pandas Libary

```python
import pandas as pd #import the pandas library to it
```

Step-2: Create a sample data

```python
data = {
    "ID": [101,102,103,104,105,106,107],
    "Name": ["Rahul","priya","Amit","Riya","Karan","Krisha","Harry"],
    "Age": [20,21,19,20,22,24,28],
    "city" : ["Surat","Navsari","Valsad","Bharuch","Surat","Gandhinagar","Ahmedabad"],
    "Marks":[85,92,78,88,81,84,90]
}
df = pd.DataFrame(data) #create a dataframe
df #dispalys the created dataset
```

Step-3: Create a dataframe

```python
df = pd.DataFrame(data) #create a dataframe
df #dispalys the created dataset
```

Step-4: Saves data as a csv file

```python
df.to_csv("student.csv",index=False) # Export to CSV without saving the default row numbers
print("csv file created successfully.")
```

Step-5: Saves data as an excel file

```python
df.to_excel("student.xlsx",index=False) # Export to excel without saving the default row numbers
print("Excel file created successfullly")
```

Step-6: Saves data as a JSON file

```python
df.to_json("student.json",orient="records",index=4) #Export to json format
print("json file created successfullly")
```
