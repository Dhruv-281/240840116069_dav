**Practical-2 | Perform data inspection, cleaning, handling, mising values duplicates rewmoval and data transformation using pandas**

```python
import pandas as pd
data={
    "ID":[101,102,103,104,104,105,106,107],
    "Name":["Rahul","Priya","Amit",None,"Riya","Ajay","Naresh","Jenil"],
    "Age":[20,21,None,20,20,22,23,28],
    "City":["Surat","Navsari","Surat","Bharuch","Bharuch","Navsari","Surat","Ahemdabad"],
    "Marks":[85,None,78,88,88,81,90,93]
}
df = pd.DataFrame(data)
df
```

Step-5 Fill missing values

```python
df["Marks"] = df["Marks"].fillna(df["Marks"].mean()) #fillna() replaces the missing values with the calculated average
#mean() for average
df
```

Fill missing *age*

```python
df["Age"] = df["Age"].fillna(df["Age"].mean()) #fillna() replaces the missing values with the calculated average
df
```

Fill Missing Names

```python
df["Name"] = df["Name"].fillna("Unknown") #fillna() replaces the missing values with the Unknown
df
```

Step-6 Detect the duplicate records

```python
df.duplicated() #Shows which rows are duplicated
```

Count the number of duplicate rows

```python
df.duplicated().sum() #Gives the count of dupliacted rows
```

Step-7 Remove the duplicate records

```python
df.drop_duplicates() #Drop the duplicate rows from the dataset
```

Rename a column

```python
df.rename(columns={"Marks":"Score"}, inplace=True) #rename():renames the name of the column ,inplace=True replaces it permanently
df
```

Step-9 Change the datatypes

Convert age column into integer

```python
df["Age"] = df["Age"].astype(int) #astype():Converts the age column's datatype into integer
df
```

Step-10 Create a new name column final score and in that add 5 grace marks to every student's existing score

```python
df["Final Score"] = df["Score"]+5 # Create a column final score and add 5 gracing marks to the existing file
df
```

Step-11: Convert text to uppercase
Convert all students names to uppercaese

```python
df["Name"] = df["Name"].str.upper() #str.upper() converts the text to uppercase
df
```

Step-12: Convert all city names to lowercase

```python
df["City"] = df["City"].str.lower()
df
```

Step-13: Remove all spaces from the city column

```python
df["City"] = df["City"].str.strip() #Romves the whitespaces(trailing/leading) from the city column
df
```

Step-14: sort the dataset

```python
df.sort_values(by="Score" ,ascending=False) #Sort the dataset in the descending order by marks
```

Step-15: Save the clean dataset

```python
df.to_csv("Cleaned_Students.csv", index=False) #Saves the dataset as Cleaned_Students.csv file
```
