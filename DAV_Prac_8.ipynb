**Create interactive visualizations using Plotly Express and Plotly Graph Objects.**

```python
#importing all the necessary libraries
import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go
```

```python
#Creating a dataset
Data = {
    "Month": ["January", "February", "March", "April", "May", "June", "July", "August", "September", "October", "November", "December"],
    "Product":["Laptop","Mobile","Tablet","Laptop","Mobile","Tablet","Laptop","Mobile","Tablet","Laptop","Mobile","Tablet"],
    "Region":["North","South","East","West","North","South","East","West","North","South","East","West"],
    "Sales":[12000,135000,150000,145000,160000,175000,180000,170000,190000,210000,230000,250000],
    "Units_Sold":[12,30,25,14,35,30,18,38,32,21,45,40],
    "Profit":[18000,27000,30000,22000,32000,35000,28000,34000,38000,33000,46000,50000]
}
#Converting dataset into dataframe
df = pd.DataFrame(Data)
df
```

```python
#Display first five records from the dataframe
df.head()
```

```python
#Display the number of rows and columns in the dataframe
df.shape
```

```python
#Display the column names and datatype
print(df.dtypes)
```

Step-3 Create an Interactive Line Chart using Plotly Express

```python
#Create an interactive line chart on how does revenue changes throughout the year using plotly express
fig = px.line(df, x="Month", y="Sales", title="Revenue Changes Throughout the Year")
fig.show()
```

Step-4 Improve the Hover Information

```python
#Create a line chart using sales over month
#Add additional columns to hover information . Students can see product,region,Units sold and profit when they move mouse over a data point
fig = px.line(df, x="Month", y="Sales", title="Revenue Changes Throughout the Year",
              hover_data=["Product", "Region", "Units_Sold", "Profit"])#hover_data is used to print the additional information
fig.show()
```

Step-5 Create an Interactive Bar chart

```python
#Group dataset according to the product
#Calculate the total sales generated by each product
df_product = df.groupby("Product")["Sales"].sum()
df_product
#Create an Interactive Bar chart for the same
fig = px.bar(df_product, x=df_product.index, y="Sales", title="Total Sales by Product")
fig.show()
```

Step-6 Interactive Scatter Plot

```python
#To analyze the realtionship between Units sold and Sales using an interactive Scatter Plot
fig = px.scatter(df, x="Units_Sold", y="Sales", title="Relationship Between Units Sold and Sales", hover_data=["Product"])
fig.show()
```

Step-7 Interactive Scatter Plot with colour

```python
#Create an interactive scatter plot with colours
fig = px.scatter(df, x="Units_Sold", y="Sales", color="Product", title="Relationship Between Units Sold and Sales", hover_data=["Product"])
fig.show()
```
