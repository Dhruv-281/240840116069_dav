```python
#importing the pandas, numpy and matplotlib.pyplot libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
#Create a datset
data = {
    "Student_ID":[101,102,103,104,105,106,107,108,109,110,111,112,113,114,115],
    "Study_Hours":[2,3,4,5,6,7,8,9,3,4,5,6,7,10,1],
    "Attendance":[65,70,75,80,82,85,90,92,68,74,78,84,88,95,40],
    "Assignment_Score":[55,60,65,70,72,78,82,88,62,67,71,76,80,95,35],
    "Final_Marks":[50,55,62,68,72,78,82,90,58,63,69,75,81,98,25]
}
#Converting dataset into dataframe
df = pd.DataFrame(data)
```

```python
#Printing first 5 records using head()
df.head()
```

```python
#Get the basic information of the dataframe
df.info()
```

```python
#Get the number of rows and columns in the dataframe
df.shape
```

```python
#Calculate the descriptive information of the dataframe
df.describe()
```

Step-4 Calculate Correlation

```python
#Calculate the correlation between study_hours and Final_Marks , we have used the sorr to find correaltion
correlation = df["Study_Hours"].corr(df['Final_Marks'])
```

```python
#Display the correlation value
print(correlation)
```

Step-5: Calculate correlation matrix

```python
#Get the correlation matrix
correlation__matrix = df.corr()
```

```python
#Display the correlation matrix
print(correlation__matrix)
```

Step-6: Visualize correlation matrix

```python
#Using the plt plot a graph
plt.figure(figsize=(8,6))
#plt.scatter used to plot the scattered chart
plt.scatter(df['Study_Hours'], df['Final_Marks'])
#Give Study hours label to x-axis
plt.xlabel('Study Hours')
#Give the Final marks label to y-axis
plt.ylabel('Final Marks')
#Give title to the graph
plt.title('Study Hours vs Final Marks scatterplot graph')
plt.grid(True)
#Display the chart
plt.show()
```

Step-7 Calculate the covarriance

```python
#Getting the covarriance using the .cov()
covarriance = df['Study_Hours'].cov(df['Final_Marks'])
print(covarriance)
```

Step:8

```python
#Calculate the covarriance matrix
covarriance_matrix = df.cov()
print(covarriance_matrix)
```

Step-9 Detect the outliers using IQR for Final_Marks

```python
#Detect the outliers using IQR for Final_Marks
outliers = []
#Using for loop for the marks below 40 and above 90
for value in df['Final_Marks']:
    if value < 40 or value > 90:
        outliers.append(value)
print(outliers)
```

```python
#Identify the outliers hwere final marks is either below 40 or above 90
outliers = df[(df['Final_Marks'] < 40) | (df['Final_Marks'] > 90)]
print(outliers)
```

Step-10:Select the records where Final_Marks are below the lower bound or above the upper bound

```python
#Select the records where Final_Marks are below the lower bound or above the upper bound
outliers = df[(df['Final_Marks'] < 40) | (df['Final_Marks'] > 90)]
print(outliers)
```

Step-11: Count the outliers

```python
#Count the outliers
outliers = df[(df['Final_Marks'] < 40) | (df['Final_Marks'] > 90)]
#len() will give the length of the outlier string which will be also its count
print(len(outliers))
```
