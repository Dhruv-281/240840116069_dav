**Practical-4:Exploratory Data Analysis using descriptive statistics and identify trends in dataset**

Step-1:Import Required Libraries

```python
#Import the pandas library
#Pandas is used for reading ,manipulating and analyzing datasets
import pandas as pd
#importing the matplotlib library for the ploting of the graph
import matplotlib.pyplot as plt
```

Step-2:Create the dataset

```python
#Creating a dataframe from the database
Data = {
    "Month": ["January", "February", "March", "April", "May", "June", "July", "August", "September", "October", "November", "December"],
    "Units_Sold": [120,135,150,145,160,175,180,170,190,210,230,250],
    "Revenue": [120000,135000,150000,145000,160000,175000,180000,170000,190000,210000,230000,250000],
    "Profit": [20000,23000,25000,24000,27000,30000,31000,29000,33000,37000,41000,45000]
}
#Creating a Dataframe from the database
df=pd.DataFrame(Data)
#Displaying the dataframe
df
```

Step-3:Inspect the dataset

```python
#Display the first five rows of the dataset
df.head()
```

```python
#display the number of rows and columns
df.shape
```

```python
#Display all the columns names
df.columns
```

```python
#Display the basic information of the dataset
df.info()
```

Step-4:Generate Descriptive Statistics

```python
#Generate discriptive statistics information for all numeric columns
df.describe()
```

Step-5 Claculate mean

```python
#Calculating the average number of units sold per month using mean()
average_units = df["Units_Sold"].mean()
#Displaying the result
print("Average Units Sold:", average_units)
```

Step-6 calculate median

```python
#Calculating the middle value of the units_sold column using the median()
median_units = df["Units_Sold"].median()
#Displaying the median value
print("Median Units Sold:", median_units)
```

Step-7 Calculte the maximum and minimum

```python
#Finding the highest number of units sold in the month using max()
maximum_units = df["Units_Sold"].max()
print("Maximum Units Sold:", maximum_units)
#Find the lowest number of units sold in the month using min()
minimum_units = df["Units_Sold"].min()
print("Minimum Units Sold:", minimum_units)
```

Step-8 Calculate Standard Deviation

```python
#Calculating the standard deviation of units sold with the help of std()
std_units = df["Units_Sold"].std()
#Display the standard deviation
print("Standard Deviation of Units Sold:", std_units)
```

Step-9 Find the month with highest sales

```python
highest_sales_index = df["Revenue"].idxmax() #idmax() gives the index also.
#retrieve the month corresponding to that index.
highest_sales_month = df.loc[highest_sales_index, "Month"] #.
#retrieve the maximum revenue
highest_revenue = df.loc[highest_sales_index, "Revenue"]
print("Month with Highest Sales:", highest_sales_month)
print("Highest Revenue:", highest_revenue)
```

Step-10 Find the month with the lowest sales

```python
#Find the index of the row having the minimum revenue
lowest_sales_index = df["Revenue"].idxmin()
#retrieve the month corresponding to that index
lowest_sales_month = df.loc[lowest_sales_index, "Month"]
lowest_revenue = df.loc[lowest_sales_index, "Revenue"]
print("Month with Lowest Sales:", lowest_sales_month)
print("Lowest Revenue:", lowest_revenue)
```

Step-11: Calculate total revenue

```python
#Calculate the total revenue generated during the year.
#Finding the total revenue using the sum()
total_revenue = df["Revenue"].sum()
#Display the total revenue
print("Total Revenue:", total_revenue)
```

Step-12 Calculate toatl profit

```python
#Calculate the total profit generated in the year
#Calculating the total profit in the year using sum()
total_profit = df["Profit"].sum()
#Display the total profit
print("Total Profit:", total_profit)
```

Step-13 Identify the sales trend

```python
from matplotlib.lines import lineStyles
#create a line chart to visualize the monthly profit
plt.figure(figsize=(10,5))
#Plot Month on the X-axis and revenue on the Y-axis
plt.plot(df["Month"], df["Revenue"], marker="o")
#Add a title to the graph
plt.title("Monthly Revenue chart.")
#Label X-axis as month
plt.xlabel("Month")
#Label Y-axis as Revenue
plt.ylabel("Revenue")
#Rotate the months for better readability
plt.xticks(rotation=30)
#Display the graph
plt.show()
```

Step-14 Analyze Profit trend

```python
#Create a linechart for monthly profit
from matplotlib.lines import lineStyles
plt.figure(figsize=(10,5))
#Plot Month on the X-axis and profit on the Y-axis
plt.plot(df["Month"], df["Profit"], marker="o")
#Add a title to the graph
plt.title("Monthly Profit chart.")
#Label X-axis as month
plt.xlabel("Month")
#Label Y-axis as Profit
plt.ylabel("Profit")
#Rotate the months for better readability
plt.xticks(rotation=30)
#Display the graph
plt.show()
```

Step-15 Compare the Revenue and Profit

```python
#Create a line chart to compare the Revenue and Profit
plt.figure(figsize=(10,5))
#Plot Revenue
plt.plot(df["Month"], df["Revenue"], marker="o", label="Revenue")
#Plot Profit
plt.plot(df["Month"], df["Profit"], marker="o", label="Profit")
#Give the title to the graph
plt.title("Revenue and Profit Comparison")
#Label the X-ais as month
plt.xlabel("Month")
#Label the Y-axis as amount
plt.ylabel("Amount")
#It indicates the lines
plt.legend()
#Rotates the month names on the X-axis
plt.xticks(rotation=30)
#Display the garph
plt.show()
```

Step-16 Identify the growth in sales

```python
#Calculate the percentage of change in revenue
#from one month to the next
df["Revenue_Growth_in_%"] = df["Revenue"].pct_change()*100
#Displaying the result
df
```

Step-17 Identify Months with increasing revenue

```python
#Filter the dataset to display only those months which have incresing revenue
increasing_sales = df[df["Revenue_Growth_in_%"]>0]
print("Months with Increasing Revenue:")
print(increasing_sales)
```
